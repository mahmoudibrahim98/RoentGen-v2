```python
import os
from pathlib import Path
import glob

base_dir = "/home/vito/ibrahimm/projects/AI4Health/notebooks/ibrahimm/Generative-Models/images/Chest_XRay/RoentGen-v2/output/2_hcn_without_uncertainty/validation_images/step_15000"

# Count images across all gpu_* subdirectories
real_count = 0
synthetic_count = 0

# Find all gpu_* directories
gpu_dirs = sorted(glob.glob(os.path.join(base_dir, "gpu_*")))

print(f"Found {len(gpu_dirs)} GPU directories")
print("-" * 60)

for gpu_dir in gpu_dirs:
    gpu_name = os.path.basename(gpu_dir)
    
    # Count synthetic images (synthetic_*.png)
    synthetic_files = glob.glob(os.path.join(gpu_dir, "synthetic_*.png"))
    synthetic_count += len(synthetic_files)
    
    # Count real images (real_*.png)
    real_files = glob.glob(os.path.join(gpu_dir, "real_*.png"))
    real_count += len(real_files)
    
    print(f"{gpu_name}: {len(synthetic_files)} synthetic, {len(real_files)} real")

print("-" * 60)
print(f"Total number of real images: {real_count}")
print(f"Total number of synthetic images: {synthetic_count}")
print(f"Total images: {real_count + synthetic_count}")
```
